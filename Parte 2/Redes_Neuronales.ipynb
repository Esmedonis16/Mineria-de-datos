```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# ==============================================================
# 1. Cargar los datos estandarizados exportados desde R
# ==============================================================

df = pd.read_csv("/content/drive/MyDrive/Proyecto parte 2/datos_proyecto.csv")

# Variables predictoras
X = df[['sexo', 'edad', 'zona', 'depto', 'delito']]

# Variable objetivo
y = df['grupo_edad']

# Codificar y
le_y = LabelEncoder()
y_encoded = le_y.fit_transform(y)

# Escalamiento
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# División train-test
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y_encoded, test_size=0.2, random_state=42
)

# ==============================================================
# 2. Definición de la red neuronal
# ==============================================================

model1 = Sequential()
model1.add(Dense(64, activation='relu', input_shape=(X_train.shape[1],)))
model1.add(Dense(32, activation='relu'))
model1.add(Dense(len(np.unique(y_encoded)), activation='softmax'))

model1.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# ==============================================================
# 3. Entrenamiento
# ==============================================================

history1 = model1.fit(
    X_train, y_train,
    validation_split=0.2,
    epochs=20,
    batch_size=256,
    verbose=1
)

# ==============================================================
# 4. Predicción para escenarios simulados
# ==============================================================

escenario1 = np.array([[1, 35, 11, 1, 1]])   # sexo, edad, zona, depto, delito
escenario1 = scaler.transform(escenario1)

pred1 = model1.predict(escenario1)
grupo_pred1 = le_y.inverse_transform([np.argmax(pred1)])

print("Predicción grupo etario:", grupo_pred1[0])
```

## Modelo 2 — Predicción de Zona

```python
# Variable objetivo zona
y2 = df['zona']
le2 = LabelEncoder()
y2_encoded = le2.fit_transform(y2)

X_train2, X_test2, y_train2, y_test2 = train_test_split(
    X_scaled, y2_encoded, test_size=0.2, random_state=42
)

model2 = Sequential()
model2.add(Dense(64, activation='relu', input_shape=(X_train2.shape[1],)))
model2.add(Dense(32, activation='relu'))
model2.add(Dense(len(np.unique(y2_encoded)), activation='softmax'))

model2.compile(optimizer='adam',
               loss='sparse_categorical_crossentropy',
               metrics=['accuracy'])

history2 = model2.fit(
    X_train2, y_train2,
    validation_split=0.2,
    epochs=20,
    batch_size=256
)

# Escenario ejemplo
esc2 = np.array([[2, 35, 2, 5, 1]])
esc2 = scaler.transform(esc2)
pred2 = model2.predict(esc2)
zona_pred = le2.inverse_transform([np.argmax(pred2)])
print("Predicción zona:", zona_pred[0])
```

## Modelo 3 — Predicción de delito

```python
# Filtrar delitos frecuentes
frecuencias = df['delito'].value_counts()
delitos_validos = frecuencias[frecuencias > 50].index

df2 = df[df['delito'].isin(delitos_validos)]

X3 = df2[['sexo', 'edad', 'grupo_edad', 'zona', 'depto']]
y3 = df2['delito']

le3 = LabelEncoder()
y3_encoded = le3.fit_transform(y3)

X_train3, X_test3, y_train3, y_test3 = train_test_split(
    X3, y3_encoded, test_size=0.2, random_state=42
)

# Escalamiento
scaler3 = StandardScaler()
X_train3 = scaler3.fit_transform(X_train3)
X_test3 = scaler3.transform(X_test3)

# Modelo
model3 = Sequential()
model3.add(Dense(128, activation='relu', input_shape=(X_train3.shape[1],)))
model3.add(Dense(64, activation='relu'))
model3.add(Dense(len(np.unique(y3_encoded)), activation='softmax'))

model3.compile(optimizer='adam',
               loss='sparse_categorical_crossentropy',
               metrics=['accuracy'])

history3 = model3.fit(
    X_train3, y_train3,
    validation_split=0.2,
    epochs=30,
    batch_size=256
)

# Escenario ejemplo
esc3 = np.array([[1, 60, 9, 1, 10]])
esc3 = scaler3.transform(esc3)

pred3 = model3.predict(esc3)
delito_pred = le3.inverse_transform([np.argmax(pred3)])

print("Predicción delito:", delito_pred[0])
```
